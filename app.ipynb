```python
import pandas as pd
import numpy as np

df_vendas = pd.read_csv('vendas_tech.csv')
```

```python
# Criando uma cópia dos dados para análise
df_analise = df_vendas
display(df_analise.head(10))
```

```python
# Criando a coluna de faturamento
df_analise['Faturamento'] = df_analise['Qtd'] * df_analise['Preco_Unitario']
```

```python
# Tratamento dos Dados

df_analise = df_analise.drop(columns=['Data_Base'])
df_analise = df_analise.dropna()
df_analise['Loja'] = df_analise['Loja'].str.strip().str.title()
df_analise['Loja'] = df_analise['Loja'].fillna('Online')
df_analise = df_analise.drop_duplicates(subset=['ID_Pedido'], keep='first')

df_analise['Data'] = pd.to_datetime(df_analise['Data'], format='%Y-%m-%d')

display(df_analise.head(10))
```

```python
# Faturamento empresa
df_faturamento_empresa = (df_analise['Qtd'] * df_analise['Preco_Unitario']).sum()
print(f'Faturamento Empresa: {df_faturamento_empresa:.2f}')
```

```python
# Produto que mais vendeu (valor e quantidade)
df_produto_mais_vendido = (df_analise.groupby('Produto')[['Faturamento', 'Qtd']].sum().sort_values('Faturamento', ascending=False))
display(df_produto_mais_vendido)
```

```python
# Criando uma coluna para região - baseado na loja
dic_regioes = {
    'São Paulo': 'Sudeste',
    'Belo Horizonte': 'Sudeste',
    'Online': 'Online',
    'Rio De Janeiro': 'Sudeste',
    'Salvador': 'Nordeste',
    'Recife': 'Nordeste',
    'Curitiba': 'Sul',
    'Porto Alegre': 'Sul'
}
df_analise['Região'] = df_analise['Loja'].map(dic_regioes)
display(df_analise.head(20))
```

```python
# Região com maior faturamento
df_regiao_maior_fat = df_analise.groupby('Região')['Faturamento'].sum().sort_values(ascending=False)
display(df_regiao_maior_fat)
```

```python
# Principais Clientes - TOP 5
df_principais_clientes = df_analise.groupby('Cliente')['ID_Pedido'].count().sort_values(ascending=False).head(5)
display(df_principais_clientes)
```

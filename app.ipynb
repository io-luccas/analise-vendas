```python
import pandas as pd
import numpy as np

df_vendas = pd.read_csv('vendas_tech.csv')
display(df_vendas)

pd.options.display.float_format = "{:.2f}".format
```

```python
display(df_vendas.head(16)) # primeiras 15 linhas da base de dados
display(df_vendas.tail(16)) # últimas 15 linhas da base de dados
display(df_vendas.sample(16)) # pega a quantidade X de linhas aleatórias da base de dados
display(df_vendas.shape) # mostra a quantidade de linhas e colunas da tabela
display(df_vendas.columns) # mostra as colunas da tabela
display(df_vendas.info()) # mostra as informações da tabela
display(df_vendas.describe()) # mostra a descrição da tabela
```

```python
#display(df_vendas.info())
display(df_vendas[['Loja', 'Cliente']]) # um colchete é uma coluna e o outro colchete é uma lista de colunas -> coluna[lista de colunas[]]
```

```python
df_analise = df_vendas.drop(columns=['Data_Base']) # Deleta uma ou mais colunas da tabela
df_analise = df_analise.dropna() # deleta as linhas que possuem pelo menos um valor em branco/nulo

df_analise['Loja'] = df_analise['Loja'].fillna('Online') # preenche os campos vazios/nulos com o valor dentro da função

display(df_analise.head(20))
```

```python
df_analise['Data'] = pd.to_datetime(df_vendas['Data'], format='%Y-%m-%d') # altera o tipo de dado para DATA - e ao lado é o formato que vai ser mostrado/alterado essa data

display(df_analise.head(20))
df_analise.info()
```

```python
# df_analise['Loja'] = df_analise['Loja'].str.strip() # remove os espaços inúteis ( antes do começo do texto, final do texto - espaços duplicados )
# df_analise['Loja'] = df_analise['Loja'].str.title() # Deixa o texto padronizado com a primeira maiúscula ( Palavra Palavra )
df_analise['Loja'] = df_analise['Loja'].str.strip().str.title() # encadeamento

display(df_analise.head(10))
```

```python
df_analise = df_analise.drop_duplicates(subset=['ID_Pedido'], keep='first') # remove linahs duplicadas ( subset = com base nessa coluna - keep = mantem o primeiro dado registrado e reomve o último )
display(df_analise.head(10))
display(df_analise.info())
```

```python
# Faturamento
df_analise['Faturamento'] = df_analise['Qtd'] * df_analise['Preco_Unitario']

# Forma de Pagamento
df_analise['Tipo_Venda'] = np.where(df_analise['Loja'] == 'Online', 'Online', 'Presencial')

# Região (usando map)
dic_regioes = {
    'São Paulo': 'Sudeste',
    'Belo Horizonte': 'Sudeste',
    'Online': 'Online',
    'Rio De Janeiro': 'Sudeste',
    'Salvador': 'Nordeste',
    'Recife': 'Nordeste',
    'Curitiba': 'Sul',
    'Porto Alegre': 'Sul'
}
df_analise['Região'] = df_analise['Loja'].map(dic_regioes)

display(df_analise.head(16))
display(df_analise.isna().sum()) # mostra o total de nulos (se houver) nas colunas
```

```python
df_faturamento_loja = df_analise.groupby('Loja')['Faturamento'].sum().reset_index()
display(df_analise.head(10))
display(df_faturamento_loja)
```

```python
df_analise = df_analise.sort_values(by=['Data', 'Faturamento']) # Ordena em ordem crescente os dados de acordo com a coluna inserida em by (pode ser mais de uma ordenação usando colchetes) - ascending = False (descrescente)
df_analise = df_analise.reset_index(drop=True) # Reseta o índice das linhas (0, 1, 2, etc) e deleta (drop=True) a coluna de índices antigo (que tinham valores como 34749, etc)
display(df_analise.head(6))
```

```python
# Loc - busca por nome da linha e da coluna = [linha, coluna]
loja = df_analise.loc[3, 'Loja']
produto = df_analise.loc[3, 'Produto']
cliente = df_analise.loc[3, 'Cliente']
print(loja, produto, cliente)

# iloc - busca por posição de índices = [índice linha, índice coluna]
loja = df_analise.iloc[3, 2]
produto = df_analise.iloc[3, 3]
cliente = df_analise.iloc[3, 6]
print(loja, produto, cliente)

# usando condicional
df_pedidos_curitiba = df_analise[df_analise['Loja']=='Curitiba']
display(df_pedidos_curitiba.head(10).sort_values(by='Faturamento', ascending=False))
```

```python
df_pedidos_curitiba.to_csv('Vendas_Curitiba.csv', index=False) # exportar planilha de vendas de Curitiba
```
